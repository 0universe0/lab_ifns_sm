```python
import numpy as np
from utils import fitPlotter, Zscore, texTabler, mytestZ
import import_ipynb
import gamma
```

```python
energyNa22   = 0.5110
energyNa22_2  = 1.2745
energyCo60_1 = 1.1732
energyCo60_2 = 1.3325
energyCs137  = 0.66166

errenergyNa22   = 0.001
errenergyNa22_2 = 0.0001
errenergyCo60_1 = 0.001
errenergyCo60_2 = 0.001
errenergyCs137  = 0.0001
```

```python
Eg = np.array([energyNa22, energyNa22_2, energyCo60_1, energyCs137]) # MeV
ErrEg = np.array([errenergyNa22, errenergyNa22_2, errenergyCo60_1, errenergyCs137])

SpExp = np.array([208, 628,570, 287])
errSpExp = np.array([16,30, 24, 16])
chSp = np.array([i for i in zip(SpExp,errSpExp)])
E_Sp, errE_Sp = gamma.calibration(chSp[:,0], chSp[:,1])
E_Sp, errE_Sp
```

```python
Esp = np.array([i for i in zip(E_Sp,errE_Sp)])

me = 0.511 # MeV
SpTh = Eg * ((Eg/me) / (0.5 + Eg/me))
ErrSpTh = (ErrEg / Eg) * SpTh

E = np.array([i for i in zip(SpTh, ErrSpTh)])

Zscore(Esp,E)
#texTabler([E_Sp, SpTh, Zscore(Esp,E)],[errE_Sp, ErrSpTh, np.ones(len(E))*0.001 ],["E_{\text{sp, exp}}", "E_{\text{sp, th}", "Z_\text{score}"])
```

```python
Eg = np.array([energyNa22, energyCo60_1, energyCo60_2, energyCs137]) # MeV
ErrEg = np.array([errenergyNa22, errenergyCo60_1, errenergyCo60_2, errenergyCs137])
chB = np.array([106,131,131,116])
echB = np.array([11,16,16,12])
Back, eBack = gamma.calibration(chB, echB)
me = 0.511 # MeV
SpTh = Eg * ((Eg/me) / (0.5 + Eg/me))
ErrSpTh = (ErrEg / Eg) * SpTh
BackTh = Eg - SpTh
ErrBackTh = (ErrEg / Eg) * BackTh
B = np.array([i for i in zip(Back,eBack)])
TB = np.array([i for i in zip(BackTh,ErrBackTh)])
Zscore(B,TB)
```
